## ★はじめにスタートボタンを押してください。

```python
# @title
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# カスタムCSSでボタン色（ダーク・ライト両対応）
custom_css = """
<style>
button.widget-button {
    background-color: #2c7be5 !important;
    color: white !important;
    font-weight: bold;
    border: none !important;
}
</style>
"""

# 塩基質量（一本鎖ベース）
base_weights = {
    'A': 313.2095,
    'T': 304.1951,
    'G': 329.2085,
    'C': 289.1846
}

AVOGADRO = 6.02214076e23

# 入力欄（フォント大きめ）
sequence_input = widgets.Textarea(
    value='',
    placeholder='ここに配列を貼り付けてください。',
    description='配列:',
    layout=widgets.Layout(width='100%', height='150px'),
    style={'description_width': '60px'}
)

concentration_input = widgets.FloatText(
    value=0.0,
    description='ng/μL:',
    style={'description_width': '60px'}
)

output_area = widgets.Output(layout=widgets.Layout(width='100%', height='180px', border='1px solid #ccc'))

# 計算関数（正しい分子量補正を含む）
def calculate_copy_number(seq, conc):
    seq = seq.replace('\n', '').replace(' ', '').upper()
    length = len(seq)
    mw = sum(base_weights.get(base, 0) for base in seq)
    g_per_uL = conc * 1e-9
    moles_per_uL = g_per_uL / (mw * 2)  # ここが重要（2倍することで二本鎖換算）
    copies_per_uL = moles_per_uL * AVOGADRO

    return f"""🧬 結果:
📏 長さ: {length} bp
⚖️ 分子量（MW）: {mw:.2f} g/mol
🧪 濃度: {conc:.1f} ng/μL
🧾 コピー数: {copies_per_uL:.3e} copies/μL"""

# 実行・クリアボタン
run_button = widgets.Button(
    description="計算",
    layout=widgets.Layout(width='100px', height='40px')
)

clear_button = widgets.Button(
    description="クリア",
    layout=widgets.Layout(width='100px', height='40px'),
    button_style='danger'
)

def on_run_clicked(b):
    with output_area:
        clear_output()
        result = calculate_copy_number(sequence_input.value, concentration_input.value)
        print(result)

def on_clear_clicked(b):
    sequence_input.value = ''
    concentration_input.value = 0.0
    with output_area:
        clear_output()

run_button.on_click(on_run_clicked)
clear_button.on_click(on_clear_clicked)

# 表示
display(HTML(custom_css))
display(widgets.HTML("<h3 style='font-size: 24px;'>🧬 DNA コピー数計算ツール</h3>"))
display(sequence_input)
display(concentration_input)
display(widgets.HBox([run_button, clear_button]))
display(output_area)
```
